```python
def bernoulli(n):
    A = [0] * (n + 1)

    for m in range(n + 1):
        A[m] = 1 / (m + 1)

        for j in range(m, 0, -1):
            A[j - 1] = j * (A[j - 1] - A[j])

    return A[0]


count = int(input("How many Bernoulli numbers should be printed? "))

bernoulli_numbers = []

for i in range(count):
    number = bernoulli(i)
    bernoulli_numbers.append(number)
    print(f"B{i} = {number:.6f}")
```
